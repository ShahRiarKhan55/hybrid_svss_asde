```python
import os, gc, warnings, numpy as np, pandas as pd, GPy, deepgp

warnings.filterwarnings("ignore")
GPy.util.linalg.jitchol_MAXTRIES = 20

DATA_ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/FPGA DATA/Raw-data_exahaustive"
OUT_ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/hybrid_svss_asde/hybrid_svss_asde_dgpr"

MODE = "all"  # "single" or "all"
SINGLE_FPGA = 1
SINGLE_PATH = 1
FPGA_START, FPGA_END, NUM_PATHS = 1, 50, 32

AL_RATIO = 0.03
SDE_RATIO = 0.07
SVTS_ALPHA = 0.5
SVTS_BETA = 0.5
K_NEIGHBORS = 2
ROUGHNESS_THRESH = 2.0
SEED = 42
MAX_ITERS = 200
NUM_INDUCING = 50
CALIB_FPGA = 1


def load_dat(fpga, path):
    grid = np.loadtxt(
        os.path.join(DATA_ROOT, f"FPGA-{fpga}", f"data{path}.dat"), delimiter=","
    )
    rows, cols = grid.shape
    rec = [
        (x, y, grid[y, x])
        for y in range(rows)
        for x in range(cols)
        if grid[y, x] != 0.0
    ]
    df = pd.DataFrame(rec, columns=["X", "Y", "Value"])
    median, std = df.Value.median(), df.Value.std()
    return df[np.abs(df.Value - median) <= 3 * std].reset_index(drop=True)


def dgpr(train_df, test_df):
    np.random.seed(SEED)
    Xtr, Ytr = train_df[["X", "Y"]].values, train_df[["Value"]].values
    Xte = test_df[["X", "Y"]].values
    xm, xs, ym, ys = Xtr.mean(0), Xtr.std(0) + 1e-9, Ytr.mean(), Ytr.std() + 1e-9
    m = deepgp.DeepGP(
        [1, 3, 2],
        (Ytr - ym) / ys,
        X=(Xtr - xm) / xs,
        num_inducing=min(NUM_INDUCING, len(Xtr)),
        back_constraint=False,
        kernels=[GPy.kern.RBF(3), GPy.kern.RBF(2)],
    )
    m.optimize(messages=False, max_iters=MAX_ITERS)
    mu, var = m.predict((Xte - xm) / xs)
    out = mu.flatten() * ys + ym, var.flatten() * ys**2
    del m
    gc.collect()
    return out


def svts_select(df):
    rng = np.random.default_rng(SEED)
    n = len(df)
    step = max(1, round(n * 0.01))
    target = max(1, round(n * AL_RATIO))
    X = df[["X", "Y"]].values.astype(float)
    Y = df["Value"].values.astype(float)
    train = list(rng.choice(n, size=step, replace=False))
    while len(train) < target:
        remaining = [i for i in range(n) if i not in train]
        m = GPy.models.GPRegression(
            X[train],
            Y[train].reshape(-1, 1),
            GPy.kern.RBF(input_dim=2),
            normalizer=True,
        )
        m.optimize(messages=False)
        mean_r, var_r = m.predict(X[remaining])
        mean_s, var_s = m.predict(X[train])
        all_mean = np.concatenate([mean_r.ravel(), mean_s.ravel()])
        all_var = np.concatenate([var_r.ravel(), var_s.ravel()])
        mean_z = (all_mean - all_mean.mean()) / (all_mean.std() + 1e-9)
        var_z = (all_var - all_var.mean()) / (all_var.std() + 1e-9)
        n_r = len(mean_r)
        feat_r = np.column_stack([mean_z[:n_r], var_z[:n_r]])
        feat_s = np.column_stack([mean_z[n_r:], var_z[n_r:]])
        centroid = feat_s.mean(axis=0)
        centroid_c = centroid - centroid.mean()
        feat_r_c = feat_r - feat_r.mean(axis=1, keepdims=True)
        sim = (feat_r_c @ centroid_c) / (
            np.linalg.norm(feat_r_c, axis=1) * np.linalg.norm(centroid_c) + 1e-9
        )
        sim_n = (sim - sim.min()) / (sim.max() - sim.min() + 1e-9)
        var_n = (var_r.ravel() - var_r.min()) / (var_r.max() - var_r.min() + 1e-9)
        scores = SVTS_ALPHA * sim_n + SVTS_BETA * var_n
        pick = min(step, target - len(train), len(remaining))
        train += [remaining[i] for i in np.argsort(scores)[::-1][:pick]]
    return train


def adaptive_sde(df, train):
    target = max(1, round(len(df) * SDE_RATIO))
    ab = {}
    for j in train:
        xj, yj = df.loc[j, "X"], df.loc[j, "Y"]
        others = sorted(
            [p for p in train if p != j],
            key=lambda p: (df.loc[p, "X"] - xj) ** 2 + (df.loc[p, "Y"] - yj) ** 2,
        )
        nn = others[:K_NEIGHBORS]
        if not nn:
            ab[j] = (1, 1)
            continue
        dx = np.mean([abs(df.loc[p, "X"] - xj) for p in nn])
        dy = np.mean([abs(df.loc[p, "Y"] - yj) for p in nn])
        bump = 1 if np.std([df.loc[p, "Value"] for p in nn]) > ROUGHNESS_THRESH else 0
        ab[j] = (int(np.ceil(dx)) + bump, int(np.ceil(dy)) + bump)
    default_ab = (
        int(np.mean([v[0] for v in ab.values()])),
        int(np.mean([v[1] for v in ab.values()])),
    )
    selected, sde = list(train), []
    for i in np.random.RandomState(SEED).permutation(df.index):
        if i in train or len(sde) >= target:
            continue
        x, y = df.loc[i, "X"], df.loc[i, "Y"]
        if all(
            abs(df.loc[j, "X"] - x) > ab.get(j, default_ab)[0]
            or abs(df.loc[j, "Y"] - y) > ab.get(j, default_ab)[1]
            for j in selected
        ):
            sde.append(i)
            selected.append(i)
            ab[i] = default_ab
    if len(sde) < target:
        leftover = [
            i
            for i in np.random.RandomState(SEED).permutation(df.index)
            if i not in train and i not in sde
        ]
        sde += leftover[: target - len(sde)]
    return sde


calib_positions = {}


def get_calib_positions(path):
    if path not in calib_positions:
        df = load_dat(CALIB_FPGA, path)
        al = svts_select(df)
        sde = adaptive_sde(df, al)
        calib_positions[path] = set(map(tuple, df.loc[al + sde, ["X", "Y"]].values))
    return calib_positions[path]


def run_path(fpga, path):
    f_out = os.path.join(
        OUT_ROOT, f"FPGA{fpga}", f"fpga{fpga}_path{path}_hybrid_svts_sde_dgpr.csv"
    )
    if os.path.exists(f_out):
        return
    df = load_dat(fpga, path)
    pos_set = get_calib_positions(path)
    train_idx = df[df.apply(lambda r: (r.X, r.Y) in pos_set, axis=1)].index.tolist()
    test_idx = [i for i in df.index if i not in train_idx]
    mu, var = dgpr(df.loc[train_idx], df.loc[test_idx])

    out = df.loc[test_idx, ["X", "Y", "Value"]].copy()
    out["Predicted"] = mu
    out["Variance"] = var

    out_dir = os.path.join(OUT_ROOT, f"FPGA{fpga}")
    os.makedirs(out_dir, exist_ok=True)
    out.to_csv(
        os.path.join(out_dir, f"fpga{fpga}_path{path}_hybrid_svts_sde_dgpr.csv"),
        index=False,
    )
    print(f"FPGA-{fpga} Path-{path}: train={len(train_idx)}/{len(df)} saved.")


if __name__ == "__main__":
    jobs = (
        [(SINGLE_FPGA, SINGLE_PATH)]
        if MODE == "single"
        else [
            (f, p)
            for f in range(FPGA_START, FPGA_END + 1)
            for p in range(1, NUM_PATHS + 1)
        ]
    )
    for fpga, path in jobs:
        run_path(fpga, path)
```

```python
import numpy as np, pandas as pd

ROOT_FOLDER = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/FPGA DATA/Raw-data_exahaustive"
OUT_ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/hybrid_svss_asde/hybrid_svss_asde_dgpr"
FPGA_START, FPGA_END = 1, 50
NUM_PATHS = 32


def load_dat(fp):
    grid = np.loadtxt(fp, delimiter=",")
    rows, cols = grid.shape
    vals = [grid[y, x] for y in range(rows) for x in range(cols) if grid[y, x] != 0.0]
    vals = pd.Series(vals)
    vals = vals[np.abs(vals - vals.median()) <= 3 * vals.std()]
    return vals


rows = []
for fpga in range(FPGA_START, FPGA_END + 1):
    for path in range(1, NUM_PATHS + 1):
        vals = load_dat(f"{ROOT_FOLDER}/FPGA-{fpga}/data{path}.dat")
        rows.append(
            {
                "FPGA": fpga,
                "PATH": path,
                "Max Value": vals.max(),
                "Min Value": vals.min(),
                "dspec": vals.max() - vals.min(),
            }
        )

dspec_df = pd.DataFrame(rows)

df = pd.read_csv(f"{OUT_ROOT}/combined_hybrid_svss_asde_dgpr.csv")
df = df.merge(dspec_df, on=["FPGA", "PATH"], how="left")
df.to_csv(f"{OUT_ROOT}/combined_hybrid_svss_asde_dgpr.csv", index=False)
```

```python
import os, numpy as np, pandas as pd

DATA_ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/FPGA DATA/Raw-data_exahaustive"
TEST_ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/hybrid_svts_sde_dgpr/FPGA1"
OUT_ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/hybrid_svts_sde_dgpr/FPGA1T"

FPGA = 1
NUM_PATHS = 32

os.makedirs(OUT_ROOT, exist_ok=True)

for path in range(1, NUM_PATHS + 1):
    # Load full raw data
    grid = np.loadtxt(
        os.path.join(DATA_ROOT, f"FPGA-{FPGA}", f"data{path}.dat"), delimiter=","
    )
    rows, cols = grid.shape
    rec = [
        (x, y, grid[y, x])
        for y in range(rows)
        for x in range(cols)
        if grid[y, x] != 0.0
    ]
    full = pd.DataFrame(rec, columns=["X", "Y", "Value"])
    median, std = full.Value.median(), full.Value.std()
    full = full[np.abs(full.Value - median) <= 3 * std].reset_index(drop=True)

    # Load test points CSV
    test_file = os.path.join(TEST_ROOT, f"fpga{FPGA}_path{path}_hybrid_svts_sde_dgpr.csv")
    if not os.path.exists(test_file):
        print(f"Path-{path}: test file not found, skipping.")
        continue
    test = pd.read_csv(test_file)[["X", "Y"]]

    # Training points = full data minus test points
    test_set = set(zip(test.X, test.Y))
    train = full[~full.apply(lambda r: (r.X, r.Y) in test_set, axis=1)]

    train.to_csv(os.path.join(OUT_ROOT, f"fpga{FPGA}_path{path}_train_hybrid_svts_sde_dgpr.csv"), index=False)
    print(f"Path-{path}: {len(train)} training points saved.")
```
