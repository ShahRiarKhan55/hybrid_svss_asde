```python
import pandas as pd, os

# EDIT THESE PATHS
METHOD_PATHS = {
    "Hybrid AL SDE": "C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/hybrid_al_sde/FPGA1T/fpga1_path{n}_train_hybrid_al_sde.csv",
    "Hybrid SVTS SDE": "C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/hybrid_svts_sde/FPGA1T/fpga1_path{n}_train_hybrid_svts_sde.csv",
    "Raw AL": "C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/raw_al/FPGA1T/fpga1_path{n}_train_raw_al.csv",
    "SVTS": "C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/svts/FPGA1T/fpga1_path{n}_train_svts.csv",
    "SDE": "C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/sde/FPGA1T/fpga1_path{n}_train_sde.csv",
}
OUT_FOLDER = "C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/common_positions"
NUM_PATHS = 32

os.makedirs(OUT_FOLDER, exist_ok=True)

for n in range(1, NUM_PATHS + 1):
    all_xy = set()
    method_sets = {}
    for method, pattern in METHOD_PATHS.items():
        df = pd.read_csv(pattern.format(n=n))
        xy = set(zip(df.X, df.Y))
        method_sets[method] = xy
        all_xy |= xy

    rows = []
    for x, y in all_xy:
        row = {"X": x, "Y": y}
        total = 0
        for method in METHOD_PATHS:
            hit = 1 if (x, y) in method_sets[method] else 0
            row[method] = hit
            total += hit
        row["Total_Selected"] = total
        rows.append(row)

    out_df = pd.DataFrame(rows).sort_values("Total_Selected", ascending=False)
    out_df.to_csv(f"{OUT_FOLDER}/FPGA1_path{n}_common_positions.csv", index=False)

print("Done")
```

```python
import pandas as pd, numpy as np, os

ROOT = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/FPGA DATA/Raw-data_exahaustive"  # folder containing FPGA-1/data{n}.dat
COMMON_FOLDER = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/common_positions"
OUT_FOLDER = r"C:/Users/sysdp/OneDrive - 島根大学/Desktop/Research/MY PAPERS/Experiment 2/FPGA0"
NUM_PATHS = 32

os.makedirs(OUT_FOLDER, exist_ok=True)

for n in range(1, NUM_PATHS + 1):
    common = pd.read_csv(f"{COMMON_FOLDER}/FPGA1_path{n}_common_positions.csv")
    common = common[common.Total_Selected >= 3].copy()

    grid = np.loadtxt(f"{ROOT}/FPGA-1/data{n}.dat", delimiter=",")
    n_active = np.count_nonzero(grid != 0.0)

    print(f"Path-{n}: train={len(common)}/{n_active} saved")

    values = [grid[int(y), int(x)] for x, y in zip(common.X, common.Y)]
    out_df = pd.DataFrame({"X": common.X, "Y": common.Y, "VALUE": values})
    out_df.to_csv(f"{OUT_FOLDER}/FPGA0_path{n}.csv", index=False)

print("Done")
```
